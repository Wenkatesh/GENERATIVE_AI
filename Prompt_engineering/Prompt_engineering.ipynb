# **Advanced Prompt Engineering**

## Zero Shot Prompting
- The zero-shot prompt directly instructs the model to perform a task without any additional examples to steer it.

```python
response = client.models.generate_content(
    model = 'gemini-2.5-flash',
    contents = "Anil travels 45 km in 3 hours. What is the speed of Anil?"
)
print(response.text)
```

## Few shot Prompting
- Provide few examples of task, along with expected output
- Model learns from examples and tries to generalize the pattern

```python
response = client.models.generate_content(
    model = 'gemini-2.5-flash-lite',
    contents = """Solve these math word problems (exact calculation style):

Q: Ram has 15 apples, gives 6 to Shyam. How many left?
A: 9 apples

Q: Sita buys 24 pens at ₹5 each. Total cost?
A: ₹120

Q: Anil travels 45 km in 3 hours. Speed?"""
)
print(response.text)
```

## Chain of Thought
- Improves reasoning capabilities
- Break down complex task into series of intermediate steps

- Prompt includes
  - initial task +
  - instruct LLM to generate step-by-step thought process
- Output
  - shows intermediate steps and
  - reasoning process
  - final answer

### Few Shot COT

```python
system_prompt = "You are a helpful assistent and answer precise and concise."

user_prompt = '''
Question: Sofia has 7 apples in her basket. Her friend Emily gives her 3 more bags of apples. Each bag contains 4 apples.
 How many apples does Sofia have in total?
step 1: calculate number of apples in the 3 bags from Emily
step 2: calculate total number of apples step 2: add results of step1 and step2
Answer: 19 apples
Question: At the bakery, there are 12 cupcakes on a tray. Sarah takes 3 cupcakes for herself.
Her friend Alex then takes half of the remaining cupcakes. How many cupcakes are left on the tray?
Answer: ??
'''

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=f"{user_prompt}, think step-by-step.",
    config=genai.types.GenerateContentConfig(
        system_instruction=system_prompt,
    )
)
print(response.text)
```

## Zero Shot COT
- we add - Please Think Step-by-Step

```python
system_prompt = "You are a helpful assistent and answer precise and concise."
user_prompt = '''Sofia has 7 apples in her basket. Her friend Emily gives her 3 more bags of apples.
Each bag contains 4 apples. How many apples does Sofia have in total?'''

response = client.models.generate_content(
    model='gemini-2.5-flash-lite',
    contents=f"{user_prompt}, think step-by-step.",
    config=genai.types.GenerateContentConfig(
        system_instruction=system_prompt,
    )
)
print(response.text)
```

## Tree of thoughts
- When humans face complex decisions, `we naturally explore multiple options`, evaluate risks, and adjust our choices when new information appears. **Tree of Thought (ToT) Prompting** brings this same flexible decision-making process to AI systems.

```python
prompt = '''
**Use the numbers 4, 9, 10, 13 to make 24 using +, −, ×, ÷.
Use each number once, parentheses allowed.**

Generate **multiple solution attempts** as follows (this is NOT chain-of-thought, but explicit task outputs):

1. Produce **five different possible expressions** (complete or partial).
2. For each expression, give a **short, non-reasoning evaluation** such as:

   * “seems promising”
   * “likely too small”
   * “possibly close to 24”
     (Do NOT explain why.)
3. Select the **two best expressions** based on those brief labels.
4. Produce **two refined versions** of each selected expression.
5. Choose the **best final expression**.
6. State whether it equals 24, and show the final expression if correct.

Return your answer in this format:

Thoughts:

* Expression 1: ..
* Expression 2: …
* Expression 3: …
* Expression 4: …
* Expression 5: …

Evaluations:

* Expression 1: promising/weak/etc.
* …

Top Selections:

* A: …
* B: …

Refinements:

* A1: …
* A2: …
* B1: …
* B2: …

Final Result:

* Best expression: …
* Equals 24?: yes/no
* Final answer (if yes): …
'''
```

```python
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
    )
print(response.text)
```
